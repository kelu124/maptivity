```python


def load_sheet(file_path,sheetname=""):
    """
    Lists all sheet names in an Excel file, then loads the second one as a DataFrame.
    
    Parameters:
        file_path (str): Path to the Excel file.
    
    Returns:
        tuple: (sheet_names, df) where
            - sheet_names is a list of all sheet names
            - df is the DataFrame of the second sheet (or None if less than 2 sheets)
    """
    # Create an ExcelFile object to inspect sheets
    xls = pd.ExcelFile(file_path)
    
    # Get list of all sheet names
    sheet_names = xls.sheet_names
    print("Sheet names found:", sheet_names)
    try:
        # Load the second sheet into a DataFrame
        df = pd.read_excel(xls, sheet_name=sheetname)
        print(f"Loaded second sheet: {sheetname}")
    except:
        return None
    return df
```

```python
dfs = []
for city in ['SHIRE', 'GOTHAM CITY', 'HOGSMEADE',"CORUSCANT"]:
    df = load_sheet("Maptivity_Cards.xlsx","SHIRE")
    df["city"]=city
    dfs.append(df)
dfs = pd.concat(dfs).reset_index(drop=True)
dfs["ID"] = dfs["city"]+"_"+dfs.index.astype(str)
dfs
```

```python
def getDesc(x,y):
    p = "Write a prompt for an image maker AI that would depict a visual for this following project:\n\n"+x+"\n\nMake sure the description matches the context of the fantasy setting of "+y+"."
    p += "\n\nAnswer only with the description, without any additional text."
    return brain.ask(p)

def getTitle(x,y):
    p = "Give a title to this project:\n\n"+x+"\n\nTry to have the title fit with the context of the fantasy setting of "+y+"."
    p += "\n\nAnswer only with the title, without any additional text; in up to 5-7 words"
    return brain.ask(p).replace('"',"").replace("**","").split(':')[-1].strip()

def getSummary(x,y):
    p = "You are a journalist and you want to write a press release or a newspaper article to this project:\n\n"+x+"\n\nTry to have the article fit with the context of the fantasy setting of "+y+"."
    p += "\n\nThe summary should be engaging and informative, highlighting the key aspects of the project.\n\n"
    p += "Avoid introductions like 'In a groundbreaking initiative,', 'The project aims to', or 'This innovative project' .. rather be friendly, using simple language a teenager could use."
    p += "\n\nAnswer only with the writing, without any additional text; in up to 200 words. Do not put styling, nor title, nor 'press release' tags for example"
    return brain.ask(p)
```

```python
dfs["desc"] = dfs.apply(lambda row: getDesc(row["Text"], row["city"]), axis=1)
dfs["title"] = dfs.apply(lambda row: getTitle(row["Text"], row["city"]), axis=1)
dfs["summary"] = dfs.apply(lambda row: getSummary(row["Text"], row["city"]), axis=1)
dfs["IDPage"] = dfs.index + 10
dfs["cityOK"] = dfs.city.apply(lambda x: x.lower().replace(" city",""))
dfs.to_parquet("maptivity_cards_with_prompts.parquet.gzip", compression="gzip", index=False)
dfs
```

```python
import qrcode

for ix, row in dfs.iterrows():
    url = "http://maptivity.online/" #"http://maptivity.online/shire_activity_7.html"
    url += str(row["cityOK"])+"_activity_"+str(row["IDPage"])+".html"
    # Create QR code
    qr = qrcode.QRCode(
        version=1,  # controls size; 1 is smallest
        error_correction=qrcode.constants.ERROR_CORRECT_H,
        box_size=10,
        border=4,
    )

    qr.add_data(url)
    qr.make(fit=True)

    # Generate image
    img = qr.make_image(fill_color="black", back_color="white")

    # Save it
    img.save("qrcodes/"+str(row["IDPage"])+".png")
```

```python
from pptx.util import Pt

def shrink_text_to_fit(shape, text, start_size=20, min_size=13):
    """
    Set text in a shape and shrink font size until it fits vertically.
    Works cross-platform (doesn't rely on fit_text()).
    """
    tf = shape.text_frame
    tf.text = text
    tf.word_wrap = True

    size = start_size
    while size >= min_size:
        # Apply the current font size to all runs
        for p in tf.paragraphs:
            for run in p.runs:
                run.font.size = Pt(size)
        # crude check: if text overflows box height, reduce size
        if tf.text == "" or tf.auto_size:
            break
        # since we can't measure rendered height easily, just reduce gradually
        size -= 2

    return size
```
