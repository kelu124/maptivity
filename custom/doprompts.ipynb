```python
import pbn_37k
from pbn_37k.ask import sigAssistant
from pptx import Presentation

import pandas as pd
from tqdm import tqdm
tqdm.pandas()

from dotenv import load_dotenv
load_dotenv()

from openai import OpenAI
import base64

from PIL import Image
import base64
from io import BytesIO
import os
from pptx import Presentation
import pandas as pd

client = OpenAI()


brain = sigAssistant(pathCache="../.cache/", llm_highend="gpt-4o",  llm_fast="gpt-4o-mini")
```

```python


def load_sheet(file_path,sheetname=""):
    """
    Lists all sheet names in an Excel file, then loads the second one as a DataFrame.
    
    Parameters:
        file_path (str): Path to the Excel file.
    
    Returns:
        tuple: (sheet_names, df) where
            - sheet_names is a list of all sheet names
            - df is the DataFrame of the second sheet (or None if less than 2 sheets)
    """
    # Create an ExcelFile object to inspect sheets
    xls = pd.ExcelFile(file_path)
    
    # Get list of all sheet names
    sheet_names = xls.sheet_names
    print("Sheet names found:", sheet_names)
    try:
        # Load the second sheet into a DataFrame
        df = pd.read_excel(xls, sheet_name=sheetname)
        print(f"Loaded second sheet: {sheetname}")
    except:
        return None
    return df
```

```python
dfs = []
for city in ['SHIRE', 'GOTHAM CITY', 'HOGSMEADE']:
    df = load_sheet("Maptivity_Cards.xlsx","SHIRE")
    df["city"]=city
    dfs.append(df)
dfs = pd.concat(dfs).reset_index(drop=True)
dfs["ID"] = dfs["city"]+"_"+dfs.index.astype(str)
dfs
```

```python
def getDesc(x,y):
    p = "Write a prompt for an image maker AI that would depict a visual for this following project:\n\n"+x+"\n\nMake sure the description matches the context of the fantasy setting of "+y+"."
    p += "\n\nAnswer only with the description, without any additional text."
    return brain.ask(p)

def getTitle(x,y):
    p = "Give a title to this project:\n\n"+x+"\n\nTry to have the title fit with the context of the fantasy setting of "+y+"."
    p += "\n\nAnswer only with the title, without any additional text; in up to 5-7 words"
    return brain.ask(p).replace('"',"").replace("**","").split(':')[-1].strip()

def getSummary(x,y):
    p = "You are a journalist and you want to write a press release or a newspaper article to this project:\n\n"+x+"\n\nTry to have the article fit with the context of the fantasy setting of "+y+"."
    p += "\n\nThe summary should be engaging and informative, highlighting the key aspects of the project.\n\n"
    p += "Avoid introductions like 'In a groundbreaking initiative,', 'The project aims to', or 'This innovative project' .. rather be friendly, using simple language a teenager could use."
    p += "\n\nAnswer only with the writing, without any additional text; in up to 200 words. Do not put styling, nor title, nor 'press release' tags for example"
    return brain.ask(p)
```

```python
dfs["desc"] = dfs.apply(lambda row: getDesc(row["Text"], row["city"]), axis=1)
dfs["title"] = dfs.apply(lambda row: getTitle(row["Text"], row["city"]), axis=1)
dfs["summary"] = dfs.apply(lambda row: getSummary(row["Text"], row["city"]), axis=1)
dfs["IDPage"] = dfs.index + 10
dfs["cityOK"] = dfs.city.apply(lambda x: x.lower().replace(" city",""))
dfs
```

```python
q = "medium"
q = "high"

q17 = """A whimsical illustration of the newly renovated municipal swimming pool in Hogsmeade, surrounded by quaint stone buildings covered in ivy. 
The pool is bustling with Hogwarts students, magical creatures, and townsfolk enjoying the refreshing water under a clear blue sky. 
Overhead, steam softly rises, indicating the newly installed heating system, while vibrant banners hang to display real-time energy savings, 
animated with magical sparkles. In the lobby, a group of excited visitors, including the headmaster and dean of hogwarts, engage in animated discussions 
about energy efficiency. Charming details like enchanted plants 
and playful house-elves add to the lively community atmosphere, highlighting the blend of practical upgrades 
and magical elements that enhance comfort and sustainability in this beloved town."""

for ix,row in dfs.iterrows():
    if not os.path.exists("images/"+row["ID"] + q+"_A4_landscape.png"):
        DESCRIPTION = row["desc"]
        if row["ID"] == "HOGSMEADE_17":
            print("Processing:", row["ID"])
            print(row["desc"])
            DESCRIPTION = q17
        if ix < 30:
            
            

            print("-- Image started for:", row["ID"])
            try:
                result = client.images.generate(
                    model="gpt-image-1",
                    prompt=DESCRIPTION,
                    size="1536x1024",
                    background="opaque",
                    quality=q,
                )
                print("-- Image generated for:", row["ID"])
                image_base64 = result.data[0].b64_json
                image_bytes = base64.b64decode(image_base64)
                with open("images/"+row["ID"] +q+".png", "wb") as f:
                    f.write(base64.b64decode(image_base64))

                # Load the image directly from bytes
                image = Image.open(BytesIO(image_bytes))

                # ---- A4 landscape ratio ----
                a4_ratio = 297 / 210  # ≈ 1.414

                # Compute new size keeping A4 ratio
                width, height = image.size
                if width / height > a4_ratio:
                    # Image is too wide → adjust width
                    new_width = int(height * a4_ratio)
                    new_height = height
                else:
                    # Image is too tall → adjust height
                    new_width = width
                    new_height = int(width / a4_ratio)

                # Resize (optional: change to a fixed A4 pixel size instead, e.g., 3508x2480)
                resized = image.resize((new_width, new_height), Image.LANCZOS)

                # Save the resized image
                resized.save("images/"+row["ID"] + q+"_A4_landscape.png", format="PNG")
            except:
                print("---- Image generation failed for:", row["ID"])
```

```python
import qrcode

for ix, row in dfs.iterrows():
    url = "http://maptivity.online/" #"http://maptivity.online/shire_activity_7.html"
    url += +str(row["cityOK"])+"_activity_"+str(row["IDPage"])+".html"
    # Create QR code
    qr = qrcode.QRCode(
        version=1,  # controls size; 1 is smallest
        error_correction=qrcode.constants.ERROR_CORRECT_H,
        box_size=10,
        border=4,
    )

    qr.add_data(url)
    qr.make(fit=True)

    # Generate image
    img = qr.make_image(fill_color="black", back_color="white")

    # Save it
    img.save("qrcodes/"+str(row["IDPage"])+".png")
```

```python

# Your data
texts = [
    "First text",
    "Second text",
    "Third text",
    "Fourth text",
    "Fifth text"
]
images = [
    "image1.jpg",
    "image2.jpg",
    "image3.jpg",
    "image4.jpg",
    "image5.jpg"
]

# Load the PowerPoint
ppt_path = "example.pptx"
prs = Presentation(ppt_path)

# Assume the first slide is the one we want to edit
slide = prs.slides[0]

# Blank layout index (usually 6)
blank_layout = prs.slide_layouts[6]

# Slide size for full-slide image
slide_width = prs.slide_width
slide_height = prs.slide_height

for text, image_path in zip(texts, images):
    # Edit text in the original slide
    for shape in slide.shapes:
        if shape.has_text_frame:
            shape.text = text
            break  # modify only the first text block

    # Add a blank slide and insert the image
    new_slide = prs.slides.add_slide(blank_layout)
    new_slide.shapes.add_picture(image_path, 0, 0, width=slide_width, height=slide_height)

# Save the updated PowerPoint
prs.save("updated_all.pptx")
```
